Model-What we should pay attention to

Gaussian Naive Bayes-Correlated thermal features violate its conditional-independence assumption; skewed features may also depart from Gaussian distributions. Still worth testing as our first baseline.

Logistic regression-Can learn linear class boundaries; correlated predictors may make coefficients less stable. Regularization will help control model complexity.

Decision tree-Can capture nonlinear thresholds and feature interactions, and does not require normally distributed features. Control depth and leaf sizes to avoid overfitting.

```python

# ============================================
# CELL 5: STRATIFIED TRAIN-VALIDATION SPLIT
# ============================================

from sklearn.model_selection import train_test_split

TARGET = "condition"
ID_COL = "unit_id"

# Predictors: exclude the target and identifier
feature_cols = [
    col for col in train_df.columns
    if col not in [TARGET, ID_COL]
]

X = train_df[feature_cols].copy()
y = train_df[TARGET].copy()

# Preserve identifiers separately for later submission
test_ids = test_df[ID_COL].copy()
X_test = test_df[feature_cols].copy()

# Split only the labelled training dataset
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training split shape:", X_train.shape)
print("Validation split shape:", X_val.shape)
print("Test dataset shape:", X_test.shape)

print("\nClass proportions in full training data:")
display(y.value_counts(normalize=True).sort_index().round(4))

print("\nClass proportions in training split:")
display(y_train.value_counts(normalize=True).sort_index().round(4))

print("\nClass proportions in validation split:")
display(y_val.value_counts(normalize=True).sort_index().round(4))

print("\nMissing values in X_train:")
display(X_train.isnull().sum())

print("\nMissing values in X_val:")
display(X_val.isnull().sum())
```

```python

# ============================================
# CELL 6: GAUSSIAN NAIVE BAYES BASELINE
# ============================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# Define feature types explicitly
numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)


# 1. Preprocess numerical features:
#    Replace missing values with training-set medians.
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])


# 2. Preprocess categorical features:
#    Impute missing categories, then one-hot encode.
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])


# 3. Apply the appropriate preprocessing to each column group.
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])


# 4. Combine preprocessing and model in one pipeline.
nb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GaussianNB())
])


# 5. Fit on the training split only.
# GaussianNB requires a dense matrix, so convert after preprocessing.
X_train_processed = preprocessor.fit_transform(X_train, y_train)
X_val_processed = preprocessor.transform(X_val)

if hasattr(X_train_processed, "toarray"):
    X_train_processed = X_train_processed.toarray()

if hasattr(X_val_processed, "toarray"):
    X_val_processed = X_val_processed.toarray()

nb_model = GaussianNB()
nb_model.fit(X_train_processed, y_train)


# 6. Predict training and validation labels.
y_train_pred_nb = nb_model.predict(X_train_processed)
y_val_pred_nb = nb_model.predict(X_val_processed)


# 7. Evaluate the baseline.
print("\n--- GAUSSIAN NAIVE BAYES BASELINE ---")
print(f"Training accuracy:   {accuracy_score(y_train, y_train_pred_nb):.4f}")
print(f"Validation accuracy: {accuracy_score(y_val, y_val_pred_nb):.4f}")

print("\nValidation classification report:")
print(
    classification_report(
        y_val,
        y_val_pred_nb,
        labels=[0, 1, 2],
        target_names=[
            "Normal",
            "Thermal Failure",
            "Mechanical Failure"
        ],
        zero_division=0
    )
)


# 8. Confusion matrix.
cm_nb = confusion_matrix(y_val, y_val_pred_nb, labels=[0, 1, 2])

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_nb,
    display_labels=[
        "Normal (0)",
        "Thermal (1)",
        "Mechanical (2)"
    ]
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Gaussian Naive Bayes — Validation Confusion Matrix")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

The training-validation gap is only −0.62 percentage points (validation is slightly higher).

The main weakness is distinguishing thermal failures. The model misses about 37% of actual thermal-failure cases, and some machines predicted as thermal failures are actually another class.
That is consistent with our EDA: several thermal features are highly correlated, while Gaussian Naive Bayes assumes conditional independence. The Gaussian assumption may also be imperfect for the skewed features.

```python

# ============================================
# CELL 7: TUNE GAUSSIAN NAIVE BAYES
# ============================================

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import accuracy_score, classification_report
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Build a fresh pipeline so preprocessing is fitted separately
# inside each cross-validation fold.
numeric_pipeline_cv = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline_cv = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor_cv = ColumnTransformer([
    ("numeric", numeric_pipeline_cv, numeric_features),
    ("categorical", categorical_pipeline_cv, categorical_features)
])

nb_pipeline_cv = Pipeline([
    ("preprocessor", preprocessor_cv),
    ("model", GaussianNB())
])

# Test smoothing values across several orders of magnitude.
param_grid = {
    "model__var_smoothing": np.logspace(-12, -3, 10)
}

cv_strategy = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

nb_search = GridSearchCV(
    estimator=nb_pipeline_cv,
    param_grid=param_grid,
    scoring="accuracy",
    cv=cv_strategy,
    n_jobs=-1,
    return_train_score=True,
    refit=True
)

# Hyperparameter selection uses only the training split.
nb_search.fit(X_train, y_train)

print("Best var_smoothing:", nb_search.best_params_["model__var_smoothing"])
print(f"Best mean CV accuracy: {nb_search.best_score_:.4f}")

# Display every tested value and its cross-validation results.
cv_results = pd.DataFrame(nb_search.cv_results_)

results_table = cv_results[[
    "param_model__var_smoothing",
    "mean_train_score",
    "mean_test_score",
    "std_test_score"
]].sort_values("mean_test_score", ascending=False)

results_table = results_table.rename(columns={
    "param_model__var_smoothing": "var_smoothing",
    "mean_train_score": "Mean CV training accuracy",
    "mean_test_score": "Mean CV validation accuracy",
    "std_test_score": "CV accuracy std"
})

display(results_table.round(5))


# Evaluate the selected model on the untouched validation split.
best_nb = nb_search.best_estimator_

y_train_pred_tuned = best_nb.predict(X_train)
y_val_pred_tuned = best_nb.predict(X_val)

print("\n--- TUNED GAUSSIAN NAIVE BAYES ---")
print(f"Training accuracy:   {accuracy_score(y_train, y_train_pred_tuned):.4f}")
print(f"Validation accuracy: {accuracy_score(y_val, y_val_pred_tuned):.4f}")

print("\nValidation classification report:")
print(classification_report(
    y_val,
    y_val_pred_tuned,
    labels=[0, 1, 2],
    target_names=["Normal", "Thermal Failure", "Mechanical Failure"],
    zero_division=0
))

cm_tuned_nb = confusion_matrix(
    y_val, y_val_pred_tuned, labels=[0, 1, 2]
)

ConfusionMatrixDisplay(
    confusion_matrix=cm_tuned_nb,
    display_labels=["Normal (0)", "Thermal (1)", "Mechanical (2)"]
).plot(cmap="Blues", values_format="d")

plt.title("Tuned Gaussian Naive Bayes — Validation")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

LOGISTIC REGRESSION (with L2 regularization)

```python

# ============================================
# CELL 8: REGULARIZED LOGISTIC REGRESSION
# ============================================

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# 1. Numerical preprocessing:
#    Impute missing values, then standardize.
numeric_pipeline_lr = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# 2. Categorical preprocessing:
#    Impute missing categories and one-hot encode.
categorical_pipeline_lr = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

# 3. Combine both preprocessing pipelines.
preprocessor_lr = ColumnTransformer([
    ("numeric", numeric_pipeline_lr, numeric_features),
    ("categorical", categorical_pipeline_lr, categorical_features)
])

# 4. Full pipeline.
lr_pipeline = Pipeline([
    ("preprocessor", preprocessor_lr),
    ("model", LogisticRegression(
        max_iter=3000,
        solver="lbfgs"
    ))
])

# 5. Tune inverse regularization strength C.
param_grid_lr = {
    "model__C": [0.01, 0.1, 1, 10, 100]
}

cv_strategy_lr = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

lr_search = GridSearchCV(
    estimator=lr_pipeline,
    param_grid=param_grid_lr,
    scoring="accuracy",
    cv=cv_strategy_lr,
    n_jobs=-1,
    return_train_score=True,
    refit=True
)

# Cross-validation uses X_train only.
lr_search.fit(X_train, y_train)

print("Best C:", lr_search.best_params_["model__C"])
print(f"Best mean CV accuracy: {lr_search.best_score_:.4f}")

# Show all tested regularization strengths.
lr_results = pd.DataFrame(lr_search.cv_results_)

lr_results_table = lr_results[[
    "param_model__C",
    "mean_train_score",
    "mean_test_score",
    "std_test_score"
]].rename(columns={
    "param_model__C": "C",
    "mean_train_score": "Mean CV training accuracy",
    "mean_test_score": "Mean CV validation accuracy",
    "std_test_score": "CV accuracy std"
}).sort_values("Mean CV validation accuracy", ascending=False)

display(lr_results_table.round(5))


# 6. Evaluate the selected model.
best_lr = lr_search.best_estimator_

y_train_pred_lr = best_lr.predict(X_train)
y_val_pred_lr = best_lr.predict(X_val)

print("\n--- TUNED LOGISTIC REGRESSION ---")
print(f"Training accuracy:   {accuracy_score(y_train, y_train_pred_lr):.4f}")
print(f"Validation accuracy: {accuracy_score(y_val, y_val_pred_lr):.4f}")

print("\nValidation classification report:")
print(classification_report(
    y_val,
    y_val_pred_lr,
    labels=[0, 1, 2],
    target_names=[
        "Normal",
        "Thermal Failure",
        "Mechanical Failure"
    ],
    zero_division=0
))

# 7. Confusion matrix.
cm_lr = confusion_matrix(y_val, y_val_pred_lr, labels=[0, 1, 2])

ConfusionMatrixDisplay(
    confusion_matrix=cm_lr,
    display_labels=[
        "Normal (0)",
        "Thermal (1)",
        "Mechanical (2)"
    ]
).plot(cmap="Blues", values_format="d")

plt.title("Logistic Regression — Validation Confusion Matrix")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

DECISION TREEES
The main risk is overfitting. We'll control that by tuning:
- max_depth: limits how many levels the tree can grow.
- min_samples_leaf: requires a minimum number of training observations in each leaf.
- criterion: we'll keep Gini impurity for this initial comparison.

```python

# ============================================
# CELL 9: DECISION TREE CLASSIFIER
# ============================================

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# 1. Decision trees do not require standardization.
numeric_pipeline_dt = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline_dt = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_dt = ColumnTransformer([
    ("numeric", numeric_pipeline_dt, numeric_features),
    ("categorical", categorical_pipeline_dt, categorical_features)
])

# 2. Full pipeline.
dt_pipeline = Pipeline([
    ("preprocessor", preprocessor_dt),
    ("model", DecisionTreeClassifier(
        criterion="gini",
        random_state=42
    ))
])

# 3. Tune tree complexity to control overfitting.
param_grid_dt = {
    "model__max_depth": [3, 5, 8, 12, None],
    "model__min_samples_leaf": [1, 5, 10, 20, 50]
}

cv_strategy_dt = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

dt_search = GridSearchCV(
    estimator=dt_pipeline,
    param_grid=param_grid_dt,
    scoring="accuracy",
    cv=cv_strategy_dt,
    n_jobs=-1,
    return_train_score=True,
    refit=True
)

# Cross-validation uses only the training split.
dt_search.fit(X_train, y_train)

print("Best parameters:", dt_search.best_params_)
print(f"Best mean CV accuracy: {dt_search.best_score_:.4f}")

# 4. Show top configurations.
dt_results = pd.DataFrame(dt_search.cv_results_)

dt_results_table = dt_results[[
    "param_model__max_depth",
    "param_model__min_samples_leaf",
    "mean_train_score",
    "mean_test_score",
    "std_test_score"
]].rename(columns={
    "param_model__max_depth": "Max depth",
    "param_model__min_samples_leaf": "Min samples per leaf",
    "mean_train_score": "Mean CV training accuracy",
    "mean_test_score": "Mean CV validation accuracy",
    "std_test_score": "CV accuracy std"
}).sort_values("Mean CV validation accuracy", ascending=False)

display(dt_results_table.head(10).round(5))


# 5. Evaluate the selected tree on the same validation split.
best_dt = dt_search.best_estimator_

y_train_pred_dt = best_dt.predict(X_train)
y_val_pred_dt = best_dt.predict(X_val)

print("\n--- TUNED DECISION TREE ---")
print(f"Training accuracy:   {accuracy_score(y_train, y_train_pred_dt):.4f}")
print(f"Validation accuracy: {accuracy_score(y_val, y_val_pred_dt):.4f}")

print("\nValidation classification report:")
print(classification_report(
    y_val,
    y_val_pred_dt,
    labels=[0, 1, 2],
    target_names=[
        "Normal",
        "Thermal Failure",
        "Mechanical Failure"
    ],
    zero_division=0
))

# 6. Confusion matrix.
cm_dt = confusion_matrix(y_val, y_val_pred_dt, labels=[0, 1, 2])

ConfusionMatrixDisplay(
    confusion_matrix=cm_dt,
    display_labels=[
        "Normal (0)",
        "Thermal (1)",
        "Mechanical (2)"
    ]
).plot(cmap="Blues", values_format="d")

plt.title("Decision Tree — Validation Confusion Matrix")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

# Find the best available fitted Random Forest
if "rf_search_fast" in globals():
    rf_model = rf_search_fast.best_estimator_
elif "best_rf_fine" in globals():
    rf_model = best_rf_fine
elif "rf_search_fine" in globals():
    rf_model = rf_search_fine.best_estimator_
elif "rf_search_regularized" in globals():
    rf_model = rf_search_regularized.best_estimator_
else:
    raise NameError(
        "No fitted Random Forest found. "
        "Please check which model variables are available."
    )

models = {"Random Forest": rf_model}

# Add other fitted models if they are available
if "dt_search_refined" in globals():
    models["Decision Tree"] = (
        dt_search_refined.best_estimator_
    )
elif "best_dt" in globals():
    models["Decision Tree"] = best_dt

if "lr_search" in globals():
    models["Logistic Regression"] = (
        lr_search.best_estimator_
    )

if "nb_search" in globals():
    models["Gaussian NB"] = nb_search.best_estimator_

print("Models available for comparison:")
for name in models:
    print("-", name)

# Convert probabilities into a consistent class order: 0, 1, 2
def get_class_probabilities(model):
    raw_probs = model.predict_proba(X_val)
    aligned_probs = np.zeros((len(X_val), 3))

    for col, class_label in enumerate(model.classes_):
        aligned_probs[:, int(class_label)] = raw_probs[:, col]

    return aligned_probs

probabilities = {
    name: get_class_probabilities(model)
    for name, model in models.items()
}

# Define candidate combinations
combinations = {
    "Random Forest only": {"Random Forest": 1}
}

if "Decision Tree" in models:
    combinations["RF + Decision Tree"] = {
        "Random Forest": 1,
        "Decision Tree": 1
    }

if "Logistic Regression" in models:
    combinations["RF + Logistic Regression"] = {
        "Random Forest": 1,
        "Logistic Regression": 1
    }

if "Decision Tree" in models and "Logistic Regression" in models:
    combinations["RF + Tree + Logistic Regression"] = {
        "Random Forest": 1,
        "Decision Tree": 1,
        "Logistic Regression": 1
    }

if "Gaussian NB" in models:
    combinations["RF + Logistic Regression + NB"] = {
        "Random Forest": 1,
        "Logistic Regression": 1,
        "Gaussian NB": 1
    }

if "Decision Tree" in models and "Logistic Regression" in models:
    combinations["RF-weighted ensemble"] = {
        "Random Forest": 3,
        "Decision Tree": 1,
        "Logistic Regression": 1
    }

if all(name in models for name in [
    "Decision Tree", "Logistic Regression", "Gaussian NB"
]):
    combinations["All four models"] = {
        "Random Forest": 3,
        "Decision Tree": 1,
        "Logistic Regression": 1,
        "Gaussian NB": 1
    }

results = []

for combo_name, weights in combinations.items():
    total_weight = sum(weights.values())

    average_probs = sum(
        probabilities[name] * weight
        for name, weight in weights.items()
    ) / total_weight

    predictions = np.argmax(average_probs, axis=1)

    results.append({
        "Combination": combo_name,
        "Validation Accuracy (%)": round(
            accuracy_score(y_val, predictions) * 100, 2
        ),
        "Macro F1 (%)": round(
            f1_score(y_val, predictions, average="macro") * 100, 2
        )
    })

results_df = pd.DataFrame(results).sort_values(
    "Validation Accuracy (%)", ascending=False
).reset_index(drop=True)

display(results_df)
```

```python

import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

# Find the best available fitted Random Forest
if "rf_search_fast" in globals():
    rf_model = rf_search_fast.best_estimator_
elif "best_rf_fine" in globals():
    rf_model = best_rf_fine
elif "rf_search_fine" in globals():
    rf_model = rf_search_fine.best_estimator_
elif "rf_search_regularized" in globals():
    rf_model = rf_search_regularized.best_estimator_
else:
    raise NameError(
        "No fitted Random Forest found. "
        "Please check which model variables are available."
    )

models = {"Random Forest": rf_model}

# Add other fitted models if they are available
if "dt_search_refined" in globals():
    models["Decision Tree"] = (
        dt_search_refined.best_estimator_
    )
elif "best_dt" in globals():
    models["Decision Tree"] = best_dt

if "lr_search" in globals():
    models["Logistic Regression"] = (
        lr_search.best_estimator_
    )

if "nb_search" in globals():
    models["Gaussian NB"] = nb_search.best_estimator_

print("Models available for comparison:")
for name in models:
    print("-", name)

# Convert probabilities into a consistent class order: 0, 1, 2
def get_class_probabilities(model):
    raw_probs = model.predict_proba(X_val)
    aligned_probs = np.zeros((len(X_val), 3))

    for col, class_label in enumerate(model.classes_):
        aligned_probs[:, int(class_label)] = raw_probs[:, col]

    return aligned_probs

probabilities = {
    name: get_class_probabilities(model)
    for name, model in models.items()
}

# Define candidate combinations
combinations = {
    "Random Forest only": {"Random Forest": 1}
}

if "Decision Tree" in models:
    combinations["RF + Decision Tree"] = {
        "Random Forest": 1,
        "Decision Tree": 1
    }

if "Logistic Regression" in models:
    combinations["RF + Logistic Regression"] = {
        "Random Forest": 1,
        "Logistic Regression": 1
    }

if "Decision Tree" in models and "Logistic Regression" in models:
    combinations["RF + Tree + Logistic Regression"] = {
        "Random Forest": 1,
        "Decision Tree": 1,
        "Logistic Regression": 1
    }

if "Gaussian NB" in models:
    combinations["RF + Logistic Regression + NB"] = {
        "Random Forest": 1,
        "Logistic Regression": 1,
        "Gaussian NB": 1
    }

if "Decision Tree" in models and "Logistic Regression" in models:
    combinations["RF-weighted ensemble"] = {
        "Random Forest": 3,
        "Decision Tree": 1,
        "Logistic Regression": 1
    }

if all(name in models for name in [
    "Decision Tree", "Logistic Regression", "Gaussian NB"
]):
    combinations["All four models"] = {
        "Random Forest": 3,
        "Decision Tree": 1,
        "Logistic Regression": 1,
        "Gaussian NB": 1
    }

results = []

for combo_name, weights in combinations.items():
    total_weight = sum(weights.values())

    average_probs = sum(
        probabilities[name] * weight
        for name, weight in weights.items()
    ) / total_weight

    predictions = np.argmax(average_probs, axis=1)

    results.append({
        "Combination": combo_name,
        "Validation Accuracy (%)": round(
            accuracy_score(y_val, predictions) * 100, 2
        ),
        "Macro F1 (%)": round(
            f1_score(y_val, predictions, average="macro") * 100, 2
        )
    })

results_df = pd.DataFrame(results).sort_values(
    "Validation Accuracy (%)", ascending=False
).reset_index(drop=True)

display(results_df)
```
